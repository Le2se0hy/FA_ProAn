# 1.1  머신러닝이란?

" 머신러닝은 데이터에서 학습하도록 컴퓨터를 프로그래밍하는 과학 "
- 공학적 정의 : 어떤 작업 T에 대한 컴퓨터 프로그램의 성능을 P를 측정했을때 경험 E로 인해 성능이 향상됐다면, 이 컴퓨터 프로그램은 작업 T와 성능 측정 P에 대해 경험 E로 학습하는 것이다.

예) 스팸필터
- 작업T = 새로운 메일이 스팸인지 구분하는 것
- 경헙E = 훈련 데이터
- 성능측정P = 직접 정의해야함. 예를 들면 정확히 분류된 메일의 비율이라면 P는 정확도

Keywords
- 훈련 세트 : 시스템이 학습하는 데 사용하는 샘플
- 훈련 사례 or 샘플 : 각각의 훈련 데이터
- 모델 : 머신러닝 시스템에서 학습하고 예측을 만드는 부분 (eg. neural network, random forest)

# 1.2 왜 머신러닝을 사용하나요?



위에서 말한 주류가 된 첫번째 머신러닝 애플리케이션인 스팸필터를 예를 들어 설명하자면

전통적인 프로그래밍 기법
1. 문제연구 - 스팸에 어떤 단어가 주로 나타나는지 살펴보며 경향 파악
2. 규칙작성 - 발견한 패턴과 경향을 감지하는 알고리즘을 작성하여 스팸인 패턴을 발견했을때 그 메일을 스팸으로 분류하게함
3. 평가 - 프로그램을 테스트하고 성능을 계속 평가
4. 오류 분석 / 론칭 - 충분한 성능이 나올때까지 오류분석 후 1단계와 2단게를 반복하고 충분한 성능이 나온다면 론칭

반면 머신러닝 기법에 기반을 둔 스팸 필터는 일반 메일에 비해 스팸에 자주 나타나는 패턴을 "**감지**"하여 어떤 단어와 구절이 스팸 메일을 판단하는 데 좋은 기준인지 "**자동으로 학습**"함.

만약 스팸메일 발송자가 '4U'를 포함한 모든 메일이 차단된다는 것을 알고 'For U'를 사용하기 시작한다면
- 전통적인 프로그래밍 : 'For U'를 포함한 메일도 스팸으로 구분하기 위한 수정이 필요함
- 머신러닝 : 사용자가 스팸으로 지정한 메일에 'For U'가 자주 나타나는 것을 **자동**으로 인식하기에 별도의 작업을 하지 않아도 이 단어를 스팸으로 분류.


전통적인 방식으론 너무 복잡하거나 알려진 알고리즘이 없는 경우에도 머신러닝을 사용한다.

또한 머신러닝 알고리즘이 학습하는 것을 알 수 있다. 모델이 예측한 결과를 통해 예상치 못한 상관관계나 새로운 추세가 발견되기도 해서 해당 문제를 더 잘 이해하도록 도와준다.
이처럼 대용량의 데이터를 분석하여 숨겨진 패턴을 발견하는 것을 **데이터 마이닝**이라고 함.

요컨대, 머신러닝은
- 기존의 전통적인 프로그래밍으로는 많은 수동적인 조정과 규칙이 필요한 문제
- 기존의 방식으로는 해결 방법이 없는 복잡한 문제
- 유동적인 환경
- 복잡한 문제와 대량의 데이터에서 인사이트를 얻기

와 같은 분야에 뛰어나기에 사용한다.

# 1.4 머신러닝 시스템의 종류

- 훈련 지도 방식 (지도, 비지도, 준지도, 자기지도, 강화학습)
- 실시간으로 점진적인 학습을 하는지의 여부 (온라인 학습, 배치 학습)
- 데이터 포인트와 새 데이터 포인트를 비교 / 훈련 데이터셋에서 패턴을 발견하여 예측 모델을 만드는지 (사례 기반 학습, 모델 기반 학습)


## 1.4.1 훈련 지도 방식

머신러닝 시스템을 학습하는 동안의 지도 형태나 정보량에 따라 분류



### [지도 학습]

지도학습에서는 알고리즙에 주입하는 훈련 데이터에 **레이블label**이라는 원하는 답이 포함됨

- 분류
- 회귀 : 특성(예측 변수 or 속성)을 사용하여 타깃 수치를 예측 (eg. 로지스틱 회귀 : 클래스에 속할 확률 출력)

### [비지도 학습]

비지도 학습은 훈련 데이터에 레이블이 없고 시스템이 아무 도움 없이 학습함
- 계층 군집 : 각 그룹을 더 작은 그룹으로 세분화하는 알고리즘
- 시각화 : 레이블이 없는 대규모의 고차원 데이터를 넣으면 도식화가 가능한 2D, 3D 표현을 만드는 알고리즘
- 차원 축소 : 상관관계가 있는 여러 특성을 하나로 합쳐 데이터를 간소화 시키는 알고리즘
- 이상치 탐지 : 학습 알고리즘에 주입하기 전 데이터셋에서 이상한 값을 자동으로 제거하는 알고리즘. 훈련하는 동안 대부분은 정상 샘플을 만나 인식하도록 훈련 후 새로운 샘플을 보고 정상 데이터인지 이상치인지 판단
- 특이치 탐지 : 훈련 세트에 있는 모든 샘플과 달라 보이는 새로운 샘플을 탐지. 알고리즘으로 감지하고 싶은 샘플을 모두 제거한 매우 깨끗한 훈련세트가 필요하다..
-> 이는 훈련세트에 이상치가 섞여있으면 모두 정상으로 인지하기에 이상치를 탐지하지 못한다. 그러므로 이상치가 없는 훈련세트가 필요하다는 뜻이다.
- 연관 규칙 학습 : 대량의 데이터에서 특성 간의 흥미로운 관계를 찾는 알고리즘


### [준지도 학습]

데이터에 레이블을 다는 것은 일반적으로 시간과 비용이 많이 들기 때문에 레이블이 없는 샘플이 많고 레이블된 샘플은 적은 경우가 많다. 일부만 레이블된 데이터를 다룸.

대부분의 준지도 학습 알고리즘은 지도 학습과 비지도 학습의 조합으로 이루어짐. 예를 들어 군집 알고리즘을 사용해 비슷한 샘플을 한 그룹으로 모으고 그 다음에 레이블이 없는 샘플에 클러스터에서 가장 많이 등장하는 레이블을 할당함. 전체 데이터셋에 레이블이 부여되고 나면 지도 학습 알고리즘을 사용할 수 있음.

### [자기 지도 학습]

레이블이 없는 데이터셋에서 레이블이 완전히 부여된 데이터셋을 생성함

자기 지도 학습의 예 : 레이블이 없는 이미지로 구성된 대량의 데이터셋이 있다면 각 이미지의 일부분을 랜덤하게 마스킹하고 모델이 원본 이미지를 복원하도록 훈련하고 훈련하는 동안 마스킹된 이미지는 모델의 입력으로 사용되며 원본 이미지는 레이블로 사용됨.





[전이 학습]
이란 한 작업에서 다른 작업으로 지식을 전달하는 것이다.

종종 자기 지도 학습을 사용해 훈련된 모델이 최종 목적이 아닌 경우가 많다. 조금 다르지만 실제 관심 대상인 작업을 위해 모델을 수정하거나 미세 튜닝함.
-> 자기 지도 학습으로 이미지 복원 모델을 훈련 후 수정하여 반려동물 분류 모델로 만듬.

### [강화 학습]

### 5.1 선형 SVM 분류

두 클래스가 직선으로 확실히 잘 나뉘어 있는데 (선형적으로 구분되어있음) 결정경계에 샘플이 가까우면 분류는 잘하겠지만 새로운 샘플엔 약할 수 있음

그 선을 도로라고 생각했을 때 도로 바깥쪽에 훈련 샘플을 더 추가해도 결정 경계에는 전혀 영향을 미치지 않음

도로 경계에 위치한 샘플에 전적으로 의지함.

→ 이런 샘플을 **서포트 벡터**라고 함


#### 5.1.1 소프트 마진 분류

- **하드 마진 분류** : 모든 샘플이 도로 바깥쪽에 올바르게 분류되어 있음

문제점
- 데이터가 선형적으로 구분될 수 있어야 제대로 작동하여 이상치에 민감
→ 이런 문제를 피하려면 좀 더 유연한 모델이 필요

도로의 폭을 가능한 넓게 유지하는 것과 마진 오류(샘플이 도로 중간이나 심지어 반대쪽에 있는 경우) 사이에 적절한 균형을 잡아야함

→ **소프트 마진 분류**

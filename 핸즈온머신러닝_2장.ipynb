### 2.2.2 성능 측정 지표 선택

**평균 제곱근 오차 (RMSE)**

![](https://drive.google.com/uc?id=18qZl7lGfIWP5iuA--uG8i2_rPGUrwRLl)

만약 이상치로 보이는 구역이 많으면 **평균 절대 오차 (MAE)**를 고려해봐야함

$MAE(𝐗,h) = 1/m \Sigma_{i=1}^m k |h(x^i)-y^i|$

RMSE와 MAE 모두 예측값의 벡터와 타깃값의 벡터 사이의 거리를 재는 방법임. 거리 측정에는 노름Norm(=방법)을 사용할 수 있음
- 유클리드 노름 : 제곱항을 합한 것의 제곱근(RMSE). $l_2$ 노름
- 맨해튼 노름 : 절댓값의 합을 계산. $l_1$ 노름
- 일반적으로 원소가 n개인 벡터 v의 $l_k$ 노름은 각 벡터의 k제곱의 합의 k분의 1 제곱으로 정의

노름의 지수가 클수록 큰 값의 원소에 치우치며 작은 값은 무시됨.
-> RMSE가 MAE보다 조금 더 이상치에 민감함

### 2.3.5 데이터 다운로드



```python
from pathlib import Path  # Path : 파일/폴더 경로를 다루기 쉽게 해주는 객체
import pandas as pd  # pandas : 데이터를 처리하고 분석해주는 라이브러리
import tarfile  # tarfile : tar파일의 압축 형식을 지원하며 압축, 해제, 파일 추가, 목록 확인 등 다양한 작업을 함
import urllib.request  # urllib.requeest : url을 열고 읽는 표준 라이브러리

def load_housing_data():  # def로 load_housing_data라는 함수를 정의
  tarball_path = Path("datasets/housing.tgz")  # datasets라는 폴더 밑에 housing.tgz라는 파일의 경로를 준비함
                                               # tarball_path는 그 경로를 가르키는 객체임
  if not tarball_path.is_file():  # .is_file을 통해 housing.tgz가 로컬에 있는지 확인. 없으면 다운로드 및 압축해제함
    Path("datasets").mkdir(parents=True, exist_ok=True)  # mkdir로 datasets라는 폴더를 만듬
                           # parents=True : datasets의 상위 경로가 주어졌을 경우 만들어줌
                           # exist_ok=True : datasets라는 폴더가 있을 경우, 에러가 안남. 해당 옵션이 없을 경우 FileExistsError가 남

    url = "https://github.com/ageron/data/raw/main/housing.tgz"  # 다운로드할 파일의 url을 문자열로 저장
    urllib.request.urlretrieve(url, tarball_path)  # urlretrieve로 url에 접속 후 파일을 다운로드 받고 tarball_path 위치에 저장
    with tarfile.open(tarball_path) as housing_tarball:  # tarfile.open으로 tarball_path를 열어서 housing_tarball이라는 핸들을 만든다.
                                                        # 여기서 housing_tarball은 tar 내부 파일 목록을 읽고, 꺼내고, 추출할 수 있는 객체가 됨
      housing_tarball.extractall(path="datasets"). # tar파일 안에 압축된 파일들을 모두 압축해제해서 datasets라는 폴더에 풀어줌
  return pd.read_csv(Path("datasets/housing/housing.csv"))  # 압축을 풀고 나면 생기는 housing.csv를 pandas로 읽어서 DataFrame으로 만들어서 반환함

housing = load_housing_data()  # 함수 실행 후 마지막 줄에 return값을 housing변수에 저장
```

### 2.3.7 테스트 세트 만들기

- **데이터 스누핑 편향** : 어떤 테스트 세트를 들여다봤을때 겉으로 드러난 어떤 패턴에 속아 특정 머신러닝 모델을 선택하게 되고, 그 테스트 세트로 일반화 오차를 추정하게 되면 매우 낙관적인 추정이 되어 기대한 성능이 나오지 않는 경우

- 테스트 세트 만들기

```python
import numpy as np

def shuffle_and_split_data(data, test_ratio):  # def로 함수 정의. 데이터셋과 테스트 세트의 비율을 매개변수로 선언
  shuffled_indices = np.random.permutation(len(data))  # 데이터 행의 개수까지의 숫자를 무작위로 섞은 배열을 변수에 넣음 (인덱스만 섞음)
  test_set_size = int(len(data) * test_ratio)  # 테스트 세트의 크기 계산 : 데이터 개수 * 테스트 세트 비율
  test_indices = shuffled_indices[:test_set_size]  # 무작위로 섞인 인덱스 중에 위에서 설정한 테스트 세트의 크기만큼의 앞부분을 테스트 세트로 사용함
  train_indices = shuffled_indices[test_set_size:]  # 나머지를 훈련 세트로 사용함
  return data.iloc[train_indices], data.iloc[test_indices]  # 각 훈련세트와 테스트세트의 데이터를 두개의 튜플로 반환함
                                                            # iloc : 인덱스의 위치로 해당 데이터를 추출하는 인덱싱 방법
```

위의 방법도 좋지만 알고리즘이 계속하여 전체 데이터셋을 봐야함

해결방법 :
- 처음 실행에서 테스트 세트를 저장하고 다음번 실행에서 불러들임
- 항상 같은 난수 인덱스가 생성되도록 np.random.permutation()을 호출하기 전에 난수 발생기의 초깃값을 괄호 속에 지정
